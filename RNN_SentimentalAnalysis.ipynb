## Pre-Processing

### 1. Converting to lowerCase

```python
df["review"] = df["review"].str.lower()
```

### 2. Removing URLs

```python
import re
```

```python
def remove_urls(text):
    text = re.sub(r"http\S+","",text) #### (pattern,replace,string)
    return text
df["review"] = df["review"].apply(remove_urls)
```

### 3. Removing Punctuations

```python
def remove_punctatutions(text):
    text = re.sub(r"[^A-Za-z0-9\s]","",text) #### remove word except->0-9 A-Z a-z
    return text

df["review"] = df["review"].apply(remove_punctatutions)
```

### 4. Removing HTML

```python
def remove_html(text):
    text = re.sub(r"<.*?>","",text) 
    return text

df["review"] = df["review"].apply(remove_html)
```

### 5. Removing the StepWords

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
# sample_text = "Hi I'm Shivam Gupta!"
# token = word_tokenize(sample_text)
# print(token)
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words('english')
    for word in tokens:
        if word in stop_words:
            text = text.replace(word,"")
            
    return text

df["review"] = df["review"].apply(remove_stopwords)
```

```python
df.head(3)
```

### 6. Stemming

```python
## running -> run
## played -> play
## PortStemming

from nltk.stem import PorterStemmer
```

```python
def stemming(text) : 
    ps = PorterStemmer()
    stemmed_words = []

    tokens = word_tokenize(text)
    for token in tokens : 
        stemmed_token = ps.stem(token)
        stemmed_words.append(stemmed_token)

    return " ".join(stemmed_words)

df["review"] = df["review"].apply(stemming)
```

## DataSets and DataLoader

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
import torch 
from torch.utils.data import TensorDataset, DataLoader
```

```python
x
```

```python
## x is a sparse matix so we convert to numpy array
```

```python
x_train = x_train.toarray()
x_test = x_test.toarray()
```

```python
train_set = TensorDataset(
    torch.from_numpy(x_train).float(),
    torch.from_numpy(y_train.values).float()
)

test_set = TensorDataset(
    torch.from_numpy(x_test).float(),
    torch.from_numpy(y_test.values).float()
)
```

```python
train_loader = DataLoader(train_set,shuffle=True,batch_size=64)
test_loader = DataLoader(test_set,shuffle=True,batch_size=64)
```

## Build RNN

```python
import torch.nn as nn
import torch.optim as optim
```

```python
class RNN(nn.Module): 
    def __init__(self, input_size, hidden_size=128, num_layers=3) :
        super().__init__()

        self.hidden_size = hidden_size
        self.num_layers = num_layers

        ## RNN layer
        self.rnn = nn.RNN(input_size, hidden_size, num_layers, batch_first=True)
        
        ## Fully Connected layer
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self,x):
        ## optional => shape(number of layer, batch size, hidden size)
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size)

        out,_ = self.rnn(x,h0)
        ## 1st value = hidden state of all the timestamp => output-> (batch,seq_len,hidden_size)
        ## 2nd value = final hidden state of last timestamp

        out = self.fc(out[:,-1,:])
        
        return out
```

```python
input_size = x_train.shape[1]

model = RNN(input_size)

criterion = nn.BCELoss() ##It is use becuase in output is will be in 2 type(Yes OR NO)
optimizer = optim.Adam(model.parameters())
```

## Training The RNN

```python
epochs = 10

for epoch in range(epochs):
    model.train()

    for xb,yb in train_loader:
        optimizer.zero_grad()

        xb=xb.unsqueeze(1) ## add singleton direction

        outputs = model(xb) ## (batch_size,1)

        outputs = torch.sigmoid(outputs.squeeze()) #(batch,size)

        loss = criterion(outputs,yb) ## compute loss
        loss.backward() ## backprop
        optimizer.step() ## weights update
        
    print(f"epoch = {epoch+1}/{epochs} and loss = {loss.item()}")
    
```

```python
## Evaluate

model.eval()

with torch.no_grad(): 
    correct_vals = 0
    total_vals = 0

    for xb,yb in test_loader:
        xb = xb.unsqueeze(1)

        outputs = model(xb)
        predicted = (torch.sigmoid(outputs.squeeze()) > 0.5).float()

        total_vals = yb.size(0)
        correct_vals = (predicted == yb).sum().item()

    print(f"Accuracy = {correct_vals/total_vals*100}")
    
```
